```python
import numpy as np
import os
import matplotlib.pyplot as plt
import pandas as pd

dane_mergesort = []
dane_quicksort = []
dane_introsort = []
```

```python
def dane_do_wykresow(nazwa_pliku) -> list:
    result = []
    with open(os.path.relpath(nazwa_pliku), 'r') as merge_file:
        tmp = merge_file.readlines()
        
        for linia in tmp:
            linia = linia.split(',')[:5]
            tmp_line = []
            for pole in linia:
                tmp_line.append(float(pole))
            tmp_line[0] = int(tmp_line[0])

            result.append(tmp_line)

    return result


dane_mergesort = dane_do_wykresow('merge.csv')
dane_quicksort = dane_do_wykresow("quick.csv")
dane_introsort = dane_do_wykresow("intro.csv")

for i in dane_mergesort:
    print(i)
```

Metoda ktora wykona wykresy dla porowania czasow danego algorytmu sortowania

```python
def zrob_wykres_porownania(tytul, wyniki):
    x_0 = [0]
    y_0 = [0]

    x_25 = [0]
    y_25 = [0]

    x_50 = [0]
    y_50 = [0]

    x_75 = [0]
    y_75 = [0]

    x_99 = [0]
    y_99 = [0]

    x_997 = [0]
    y_997 = [0]

    x_odwrotnie = [0]
    y_odwrotnie = [0]


    for rekord in wyniki:
        if rekord[1] == 0.0:
            x_0.append(rekord[0])
            y_0.append(rekord[2])
        elif rekord[1] == 0.25:
            x_25.append(rekord[0])
            y_25.append(rekord[2])
        elif rekord[1] == 0.5:
            x_50.append(rekord[0])
            y_50.append(rekord[2])
        elif rekord[1] == 0.75:
            x_75.append(rekord[0])
            y_75.append(rekord[2])
        elif rekord[1] == 0.99:
            x_99.append(rekord[0])
            y_99.append(rekord[2])
        elif rekord[1] == 0.997:
            x_997.append(rekord[0])
            y_997.append(rekord[2])
        elif rekord[1] == -1:
            x_odwrotnie.append(rekord[0])
            y_odwrotnie.append(rekord[2])
            
    plt.figure(figsize=(10,5))
    # Tworzenie wykresu
    plt.plot(x_0, y_0, marker='x', label="Nieposortowane")
    plt.plot(x_25, y_25, marker='x', label="25% posortowanych")
    plt.plot(x_50, y_50, marker='x', label="50% posortowanych")
    plt.plot(x_75, y_75, marker='x', label="75% posortowanych")
    plt.plot(x_99, y_99, marker='x', label="99% posortowanych")
    plt.plot(x_997, y_997, marker='x', label="99.7% posortowanych")
    plt.plot(x_odwrotnie, y_odwrotnie, marker='x', label="Posortowane odwrotnie")

    # Dodawanie tytułu i etykiet osi (opcjonalnie, ale to dobra praktyka)
    plt.title(tytul)
    plt.xlabel("Liczba elementow")
    plt.ylabel("Czas sortownia [ms]")
    plt.xlim(0, 1000000)
    plt.ylim(0,1000)
    plt.legend()

    # Wyświetlenie wykresu
    plt.show()
```

Wykres 1 czasy dla algorytmu MergeSort

```python
zrob_wykres_porownania("Wykres sortowania przez scalanie", dane_mergesort)
```

```python
zrob_wykres_porownania("Wykres sortowania szybkiego", dane_quicksort)
```

```python
zrob_wykres_porownania("Wykres sortowania introspektywnego", dane_introsort)
```

```python
def porownaj_sortowania(wyniki_1, wyniki_2, wyniki_3, procent = 0.00):
    x_m = [0]
    y_m = [0]

    x_q = [0]
    y_q = [0]

    x_i = [0]
    y_i = [0]

    for rekord in wyniki_1:
        if rekord[1] == procent:
            x_m.append(rekord[0])
            y_m.append(rekord[2])

    for rekord in wyniki_2:
        if rekord[1] == procent:
            x_q.append(rekord[0])
            y_q.append(rekord[2])

    for rekord in wyniki_3:
        if rekord[1] == procent:
            x_i.append(rekord[0])
            y_i.append(rekord[2])

    plt.figure(figsize=(10,5))
    # Tworzenie wykresu
    plt.plot(x_m, y_m, marker='x', label="MergeSort")
    plt.plot(x_q, y_q, marker='x', label="QuickSort")
    plt.plot(x_i, y_i, marker='x', label="IntroSort")

    # Dodawanie tytułu i etykiet osi (opcjonalnie, ale to dobra praktyka)
    if procent != -1:
        plt.title(f"Porownanie czasu sortowania dla {round(procent*100,1)}% posortowanych elemetow")
    else:
        plt.title(f"Porownanie czasu sortowania dla odwrotnie posortowanych elemetow")
    plt.xlabel("Liczba elementow")
    plt.ylabel("Czas sortownia [ms]")
    plt.xlim(0, 1000000)
    plt.ylim(0,1000)
    plt.legend()

    # Wyświetlenie wykresu
    plt.show()
```

```python
porownaj_sortowania(dane_mergesort, dane_quicksort, dane_introsort, 0.00)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Definiujemy kategorie na osi X
kategorie = ['0%','25%','50%','75%','95%', '99%', '99.7%', 'Odwrotnie']

# Tworzymy listę pozycji na osi X dla naszych kategorii: [0, 1, 2]
x = np.arange(len(kategorie)) 

# Ustawiamy szerokość pojedynczego słupka
szerokosc = 0.25

# 2. Przykładowe dane - czasy sortowania (np. dla N = 500 000)
czasy_alg1 = []  # Np. Merge Sort
czasy_alg2 = []  # Np. Quick Sort
czasy_alg3 = []  # Np. Intro Sort 

for rekord in dane_mergesort:
    if rekord[0] == 100000:
        czasy_alg1.append(rekord[2])

for rekord in dane_quicksort:
    if rekord[0] == 100000:
        czasy_alg2.append(rekord[2])

for rekord in dane_introsort:
    if rekord[0] == 100000:
        czasy_alg3.append(rekord[2])

czasy_alg2.append(10000)



# 3. Rozpoczynamy rysowanie
plt.figure(figsize=(10, 6))

# Rysujemy słupki, przesuwając je względem środka (x)
# Algorytm 1: przesuwamy w lewo o 'szerokosc'
plt.bar(x - szerokosc, czasy_alg1, width=szerokosc, label='MergeSort', color='skyblue')

# Algorytm 2: rysujemy dokładnie na środku (x)
plt.bar(x, czasy_alg2, width=szerokosc, label='QuickSort', color='lightgreen')

# Algorytm 3: przesuwamy w prawo o 'szerokosc'
plt.bar(x + szerokosc, czasy_alg3, width=szerokosc, label='IntroSort', color='salmon')

# 4. Dodajemy opisy osi, tytuł i legendę
plt.title("Porównanie czasów działania algorytmów dla skrajnych przypadków (N = 500 000)")
plt.ylabel("Czas wykonania [ms]")

# Zastępujemy liczby na osi X nazwami naszych kategorii
plt.xticks(x, kategorie)

plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7) # Dodajemy poziomą siatkę dla lepszej czytelności
plt.ylim(top=100)

# Wyświetlenie wykresu
plt.show()
```
